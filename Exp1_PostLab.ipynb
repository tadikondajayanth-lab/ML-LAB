```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
sns.set_style("whitegrid")
df = sns.load_dataset("titanic")
print("First 5 Rows:")
print(df.head())
print("\nDataset Shape:")
print(df.shape)
print("\nColumn Names:")
print(df.columns)
print("\nDataset Information:")
print(df.info())
print("\nSummary Statistics:")
print(df.describe())
print("\nMissing Values:")
print(df.isnull().sum())
plt.figure(figsize=(6,4))
sns.countplot(x="survived", data=df)
plt.title("24EU02055-Survival Count")
plt.show()
plt.figure(figsize=(6,4))
sns.countplot(x="sex", data=df)
plt.title("24EU02055-Gender Distribution")
plt.show()
plt.figure(figsize=(6,4))
sns.countplot(x="sex", hue="survived", data=df)
plt.title("24EU02055-Survival by Gender")
plt.show()
plt.figure(figsize=(6,4))
sns.countplot(x="pclass", data=df)
plt.title("24EU02055-Passenger Class Distribution")
plt.show()
plt.figure(figsize=(6,4))
sns.countplot(x="pclass", hue="survived", data=df)
plt.title("24EU02055-Survival by Passenger Class")
plt.show()
plt.figure(figsize=(8,5))
sns.histplot(df["age"].dropna(), bins=30, kde=True)
plt.title("24EU02055-Age Distribution")
plt.xlabel("Age")
plt.show()
plt.figure(figsize=(8,5))
sns.histplot(df["fare"], bins=30, kde=True)
plt.title("24EU02055-Fare Distribution")
plt.xlabel("Fare")
plt.show()
plt.figure(figsize=(8,5))
sns.boxplot(x="pclass", y="age", data=df)
plt.title("24EU02055-Age by Passenger Class")
plt.show()
plt.figure(figsize=(10,8))
numeric_df = df.select_dtypes(include=np.number)
sns.heatmap(numeric_df.corr(), annot=True, cmap="coolwarm")
plt.title("24EU02055-Correlation Heatmap")
plt.show()
sns.pairplot(df[["survived", "age", "fare", "pclass"]].dropna(), hue="survived")
plt.show()
print("\nEDA Completed Successfully!")
```
